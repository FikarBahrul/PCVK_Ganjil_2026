STUDI KASUS 1

```python
#Buat fungsi label adaptif
def tambah_label(img, nama):
    hasil = img.copy()

    h, w = hasil.shape[:2]

    # Ukuran bar berdasarkan tinggi gambar
    bar_h = int(h * 0.12)

    # Ketebalan berdasarkan ukuran gambar
    thickness = max(1, int(min(h, w) * 0.01))

    # Ukuran tulisan berdasarkan tinggi gambar
    font_scale = max(0.5, h / 700)

    # Posisi teks berdasarkan ukuran gambar
    text_x = int(w * 0.03)
    text_y = int(h * 0.08)

    # Membuat bar di bagian atas
    cv.rectangle(
        hasil,
        pt1=(0, 0),
        pt2=(w, bar_h),
        color=(0, 0, 0),
        thickness=-1
    )

    # Menambahkan nama
    cv.putText(
        hasil,
        text=nama,
        org=(text_x, text_y),
        fontFace=cv.FONT_HERSHEY_SIMPLEX,
        fontScale=font_scale,
        color=(255, 255, 255),
        thickness=thickness,
        lineType=cv.LINE_AA
    )

    return hasil
```

```python
img1 = cv.imread('foto1.jpg')

hasil1 = tambah_label(
    img1,
    'Fikar Bahrul Santoso'
)

plt.imshow(cv.cvtColor(hasil1, cv.COLOR_BGR2RGB))
plt.show()
```

```python
img2 = cv.imread('foto2.jpg')

hasil2 = tambah_label(
    img2,
    'Fikar Bahrul Santoso'
)

plt.imshow(cv.cvtColor(hasil2, cv.COLOR_BGR2RGB))
plt.show()
```

```python
#fungsi thumbnail persegi
def buat_thumbnail(img, nama_toko):
    h, w = img.shape[:2]

    # Sisi canvas mengikuti sisi terpanjang gambar
    side = max(h, w)

    # Membuat canvas persegi
    canvas = np.zeros(
        shape=(side, side, 3),
        dtype=np.uint8
    )

    # Membuat canvas berwarna
    canvas[:, :] = (240, 240, 240)

    # Menghitung posisi agar gambar berada di tengah
    x = (side - w) // 2
    y = (side - h) // 2

    # Menempatkan gambar di tengah canvas
    canvas[y:y+h, x:x+w] = img

    # Ukuran bar logo berdasarkan ukuran canvas
    bar_h = int(side * 0.10)

    # Membuat bar di bagian bawah
    cv.rectangle(
        canvas,
        pt1=(0, side - bar_h),
        pt2=(side, side),
        color=(0, 0, 0),
        thickness=-1
    )

    # Ukuran teks menyesuaikan canvas
    font_scale = max(0.5, side / 800)

    thickness = max(
        1,
        int(side * 0.008)
    )

    text_x = int(side * 0.03)
    text_y = int(side - bar_h * 0.30)

    # Menambahkan nama toko
    cv.putText(
        canvas,
        text=nama_toko,
        org=(text_x, text_y),
        fontFace=cv.FONT_HERSHEY_SIMPLEX,
        fontScale=font_scale,
        color=(255, 255, 255),
        thickness=thickness,
        lineType=cv.LINE_AA
    )

    return canvas
```

```python
#Potrait
img_portrait = cv.imread('foto_potrait.jpg')

thumbnail_portrait = buat_thumbnail(
    img_portrait,
    'TOKO FIKAR'
)

plt.imshow(
    cv.cvtColor(
        thumbnail_portrait,
        cv.COLOR_BGR2RGB
    )
)

plt.show()
```

```python
#Landscape
img_landscape = cv.imread('foto_landscape.jpg')

thumbnail_landscape = buat_thumbnail(
    img_landscape,
    'TOKO FIKAR'
)

plt.imshow(
    cv.cvtColor(
        thumbnail_landscape,
        cv.COLOR_BGR2RGB
    )
)

plt.show()
```
